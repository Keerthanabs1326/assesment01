Q21. Indian Income Tax Calculator (New Regime, FY 2024–25). Take annual income as input.Compute tax using these slabs:Up to ₹3,00,000 : Nil₹3,00,001 – ₹7,00,000 : 5%₹7,00,001 – ₹10,00,000 : 10%₹10,00,001 – ₹12,00,000 : 15%₹12,00,001 – ₹15,00,000 : 20%Above ₹15,00,000 : 30%Tax must be computed slab-wise, NOT flat. Example: income ₹8,00,000 → 0 on first 3L + 5% on next 4L + 10% onlast 1L = ₹20,000 + ₹10,000 = ₹30,000. Print breakdown clearly.

```python
inc = float(input())
slabs = [(300000, 0.0), (400000, 0.05), (300000, 0.10), (200000, 0.15), (300000, 0.20)]
tax, rem = 0, inc

for cap, rate in slabs:
    taxable = min(rem, cap)
    tax += taxable * rate
    rem -= taxable
    if rem <= 0: break
if rem > 0:
    tax += rem * 0.30

print(f"Total Tax: ₹{tax:.2f}")
```

Q31. Find all Armstrong numbers between 1 and 9999. (An n-digit number is Armstrong if the sum ofits digits each raised to the power n equals the number itself. Example: 153 = 1³ + 5³ + 3³.) Print themin a clean comma-separated list with a count at the end.

```python
res = []
for i in range(1, 10000):
    t, s, d = i, 0, len(str(i))
    while t > 0:
        s += (t % 10) ** d
        t //= 10
    if s == i:
        res.append(i)
print(", ".join(map(str, res)), f"\nCount: {len(res)}")
```

Q33. Prime check, optimized. Take a number N. Check if it is prime by iterating only up to √N. Then,print the SUM of all primes between 2 and N (inclusive). Do not use any library.

```python
n = int(input())
def is_prime(x):
    if x < 2: return False
    for i in range(2, int(x ** 0.5) + 1):
        if x % i == 0: return False
    return True

print("Prime" if is_prime(n) else "Not Prime")
print("Sum of primes:", sum(x for x in range(2, n + 1) if is_prime(x)))
```

Q45. Caesar Cipher Encryptor. Take a message and a shift value N as input. Encrypt by shifting each
letter forward by N positions in the alphabet (wrapping around, Z + 1 = A). Preserve case. Keep digits,
spaces, and special characters UNCHANGED. Example: "Hello, India!" with shift 3 → "Khoor, Lqgld!"

```python
msg, n = input("Message: "), int(input("Shift: "))
out = []
for ch in msg:
    if ch.isupper():
        out.append(chr((ord(ch) - 65 + n) % 26 + 65))
    elif ch.islower():
        out.append(chr((ord(ch) - 97 + n) % 26 + 97))
    else:
        out.append(ch)
print("".join(out))
```

Q46. Write a function `calculate_gst(amount, gst_rate=18, gst_inclusive=False)`:
- If gst_inclusive=False: amount is base, add GST on top
- If gst_inclusive=True: amount already includes GST, extract it
- Return a tuple: (base_amount, gst_amount, total_amount)
- All values rounded to 2 decimals
Test cases:
 calculate_gst(1000) → (1000.00, 180.00, 1180.00)
 calculate_gst(1180, gst_inclusive=True) → (1000.00, 180.00, 1180.00)
 calculate_gst(500, gst_rate=5) → (500.00, 25.00, 525.00)

```python
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    if gst_inclusive:
        base = amount / (1 + gst_rate / 100)
        gst = amount - base
        return round(base, 2), round(gst, 2), round(amount, 2)
    gst = amount * (gst_rate / 100)
    return round(amount, 2), round(gst, 2), round(amount + gst, 2)

print(calculate_gst(1000))
```

Q47. Write a RECURSIVE function `power(base, exp)` that computes base raised to exp, WITHOUT
using the `**` operator and WITHOUT using `pow()`. Handle exp = 0 (return 1) and negative exp
(return 1/power(base, -exp)). Test with: power(2, 10), power(5, 0), power(3, -2).

```python
def power(base, exp):
    if exp == 0: return 1
    if exp < 0: return 1 / power(base, -exp)
    return base * power(base, exp - 1)
print(power(2, 10))
```

Q48. Identity Matrix Manipulations. Using NumPy: (a) create a 5×5 identity matrix using np.eye(), (b)
multiply every element by 10, (c) transpose it (yes, it looks the same, explain in a comment WHY
transpose of an identity is itself), (d) compute the sum of diagonal elements, (e) compute the sum of
all elements. Print each step.

```python
import numpy as np
eye = np.eye(5)                  # (a)
eye_10 = eye * 10                 # (b)
trans = eye_10.T                  # (c) Transpose equals itself because symmetric (M[i,j] == M[j,i])
diag_sum = np.trace(eye_10)       # (d) 50.0
total_sum = eye_10.sum()          # (e) 50.0
print(eye, eye_10, trans, diag_sum, total_sum, sep="\n")
```

Q49. Zomato Weekly Sales Analytics. Create a NumPy array of 7-day Zomato Bengaluru sales (in ₹):
sales = np.array([12000, 18500, 14200, 22100, 19800, 35200, 28400])
[Monday, Tuesday, Wednesday, Thursday, Friday, Saturday, Sunday]
Without using loops, compute and print:
(a) Total weekly sales
(b) Daily average
(c) The day (Mon/Tue/...) with maximum sales
(d) Number of days with sales above the weekly average
(e) Percentage growth from Monday to Sunday
(f) An array of day-on-day % changes (length 6)

```python
import numpy as np

sales = np.array([12000, 18500, 14200, 22100, 19800, 35200, 28400])
days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])

print("Total:", sales.sum())
print("Average:", sales.mean())
print("Max day:", days[sales.argmax()])
print("Days > Avg:", (sales > sales.mean()).sum())
print("Mon->Sun growth %:", ((sales[-1] - sales[0]) / sales[0]) * 100)
print("DoD % change:", np.diff(sales) / sales[:-1] * 100)
```
